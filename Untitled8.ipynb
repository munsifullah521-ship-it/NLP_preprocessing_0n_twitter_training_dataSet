Lowercasing: text.lower() sab letters chhotay kar deta hai, taake “Good” aur “good” ek hi word samjha jaye.
Extra spaces hatana: re.sub("\s+"," ",text) ek se zyada spaces ko ek space bana deta hai.
Numbers hatana: re.sub("\d+"," ",text) text se saare digits nikal deta hai.
Punctuation hatana: string.punctuation se ! , . ? jaise sab symbols hat jaate hain.
Stopwords removal: stopwords.words("english") se is, the, a, and jaise aam words nikal diye jaate hain, kyunke inka koi khaas matlab nahi hota.

```python
def text_preprocessor(text):
  text = text.lower()
  text = re.sub("\s+"," ",text)
  text = re.sub("\d+"," ",text)
  NO_punch_text="".join([a for a in text if a not in string.punctuation])
  text = " ".join([a for a in NO_punch_text.split() if a not in stopwords.words("english")])

  return text
```

```python
import re
import string
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
def text_preprocessor(text):
  if not isinstance(text, str):
    return ""
  text = text.lower()
  text = re.sub(r"\s+", " ", text)
  text = re.sub(r"\d+", " ", text)
  NO_punch_text = "".join([a for a in text if a not in string.punctuation])
  text = " ".join([a for a in NO_punch_text.split() if a not in stopwords.words("english")])
  return text
```

```python
data["Reviews"] = data["Reviews"].apply(text_preprocessor)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tok = Tokenizer()
tok.fit_on_texts(data["Reviews"])
data["Reviews"] = tok.texts_to_sequences(data["Reviews"])
```
